```python
df.isnull().sum()
# for null values we will put median and we put it by each category for better results
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create colummn for age groups
labels = ['young_adult','adult','middle_age','senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels =labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchase frequency days

freq = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(freq)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
